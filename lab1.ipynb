### Завдання 1: Створити множину з 10 цілих чисел та вивести її на екран.

```python
numbers_set = {5, 12, 34, 7, 89, 21, 10, 45, 3, 77}
print("Множина з 10 цілих чисел:", numbers_set)
```

### Завдання 2: Визначити кількість елементів множини за допомогою відповідної функції.

```python
my_set = {10, 20, 30, 40, 50}
count = len(my_set)
print("Задана множина:", my_set)
print("Кількість елементів у множині:", count)
```

### Завдання 3: Створити множину. Додати до неї два нові елементи та видалити один елемент. Вивести отриману множину.

```python
my_set = {1, 2, 3, 4, 5}
print("Початкова множина:", my_set)
my_set.add(6)
my_set.add(7)
my_set.discard(3)
print("Отримана множина:", my_set)
```

### Завдання 4: Задано множину чисел і деяке число х. Визначити, чи належить число х заданій множині.

```python
numbers = {15, 25, 35, 45, 55}
x = 35
if x in numbers:
    print(f"Число {x} належить заданій множині.")
else:
    print(f"Число {x} не належить заданій множині.")
```

### Завдання 5: Задано список чисел. Створити на його основі множину та визначити, скільки різних чисел містилося у списку.

```python
numbers_list = [10, 20, 10, 30, 40, 20, 50, 10]
print("Список чисел:", numbers_list)
unique_numbers_set = set(numbers_list)
print("Множина унікальних чисел:", unique_numbers_set)
print("Кількість різних чисел у списку:", len(unique_numbers_set))
```

### Завдання 6: Задано дві множини А і В. Знайти та вивести множину, що містить усі елементи обох множин.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
union_set = A | B
print("Множина A:", A)
print("Множина B:", B)
print("Об'єднання множин (всі елементи):", union_set)
```

### Завдання 7: Задано дві множини чисел. Знайти елементи, які одночасно належать обом множинам.

```python
set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}
intersection_set = set1 & set2
print("Спільні елементи обох множин:", intersection_set)
```

### Завдання 8: Задано дві множини А і В. Визначити елементи, які належать множині А, але не належать множині В.

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
difference_set = A - B
print("Елементи, які належать множині А, але не належать множині В:", difference_set)
```

### Завдання 9: Задано текст. Створити множину всіх літер, що зустрічаються у тексті, без урахування повторень. Визначити кількість різних літер.

```python
text = "hello world"
letters_set = set(text.replace(" ", ""))
print("Множина унікальних літер:", letters_set)
print("Кількість різних літер:", len(letters_set))
```

### Завдання 10: Задано два списки чисел. За допомогою множин визначити спільні, унікальні та окремі елементи.

```python
list1 = [1, 2, 3, 4, 5, 5, 2]
list2 = [4, 5, 6, 7, 8, 4]

set1 = set(list1)
set2 = set(list2)

print("Множина 1:", set1)
print("Множина 2:", set2)
print("Спільні елементи:", set1 & set2)
print("Тільки в першому списку:", set1 - set2)
print("Тільки в другому списку:", set2 - set1)
print("Усі унікальні елементи:", set1 | set2)
```
